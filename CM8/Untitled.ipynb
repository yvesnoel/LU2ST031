```python
def suite_la_plus_longue(ma_liste):
    try :
        long=1
        longmax=1
        precedent=ma_liste[0]
    
        for elem in ma_liste:
            if elem >= precedent:
                long+=1
            else :
                longmax=max(long,longmax)
                long=1
            precedent=elem
        longmax=max(long,longmax)
    except :
        longmax=0
    return longmax
```

```python
l = [2]
print(suite_la_plus_longue(l))
```

```python
def lepluscher(a,p):
    im=0
    for i in range(len(a)):
        im=i if p[i]>p[im] else im
    return a[im]
```

```python
import numpy as np
def adroite(A,B,C):
    A=np.array(A)
    B=np.array(B)
    C=np.array(C)
    AB=B-A
    AC=C-A
    return (AB[0]*AC[1]-AB[1]*AC[0])<0
def extrait2(N,i,j):
    S=[]
    for k in range(len(N)):
        if k!= i and k!= j :
            S.append(N[k])
    return S
```

```python
adroite([1,1],[2,2],[1.5,1.5])
```

```python
import matplotlib.pyplot as plt
import random
N=[]
for i in range(10):
    N.append([random.random()*10,random.random()*10])
```

```python
def estbord(N,i,j):
    N2=extrait2(N,i,j)
    cote=adroite(N[i],N[j],N2[0])
    for k in range(1,len(N2)):
        if cote != adroite(N[i],N[j],N2[k]):
            return False
    return True
```

```python
Ab=[x[0] for x in N]
Or=[x[1] for x in N]
i,j=1,3
plt.scatter(Ab,Or)
plt.scatter([Ab[i],Ab[j]],[Or[i],Or[j]],color='red')
estbord(N,i,j)
```

```python
def corrige (liste,a,b):
    for i in range (1,len(liste)):
        liste_corrige=[]
        if liste[i]==a or liste[i]==b:
            liste_corrige.append(liste[i])
        else: 
            liste_corrige.append(liste[i-1])
            
    return liste_corrige
            
```

```python
def compte_voitures(LV,temps):
    dureevert,dureeorange,dureerouge=temps
    periode= dureevert+ dureeorange+ dureerouge
    counts= [0,0,0]
    for t in LV:
        temps_duree= t%periode
        if temps_duree <dureevert:
            counts[0]+=1
        elif temps_duree< dureevert + dureeorange:
            counts[1] +=1
        else:
            counts[2]+=1
    return counts
```

```python
V=[44,174,200,420,702,883,940,1071,1084,1098,1117,1156,1181,1196]
print(compte_voitures(V,[150,30,120]))
```

```python
def summean (liste_de_listes):
    somme=0
    compteur=0
    non_nombres=0
    for sous_liste in liste_de_listes:
        for element in sous_liste:
            if isinstance (element,(int,float)):
                somme+=element
                compteur+=1
            else:
                non_nombres+=1
    moyenne= round(somme/compteur,2) 
    return somme,moyenne,non_nombres
```
